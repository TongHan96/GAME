```python
import torch
import pandas as pd
import numpy as np
from sklearn.decomposition import PCA
```

```python
def reduce_dimension(emb: pd.DataFrame, wanted_dim: int) -> pd.DataFrame:
    pca = PCA(n_components=wanted_dim)
    reduced_emb = pca.fit_transform(emb)
    return pd.DataFrame(reduced_emb)
```

```python
emb = pd.read_csv('https://han-attention.s3.amazonaws.com/input/drug_side_concat/concat_old_12_09.csv')
emb = emb.iloc[:,1:]
reduced_emb = reduce_dimension(emb, 768)
reduced_emb.to_csv('/home/hat127/feature_selection/RD_concat_emb/RD_old_1214.csv', index=None)
```
